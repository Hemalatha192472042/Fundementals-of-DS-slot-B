```python
import pandas as pd
import matplotlib.pyplot as plt


df = pd.read_excel("education_dataset.xlsx")


print(df.head())
print(df.info())
print(df.describe())
print(df.isnull().sum())


plt.scatter(df["Attendance"], df["Internal_Marks"])
plt.title("Attendance vs Internal Marks")
plt.xlabel("Attendance")
plt.ylabel("Internal Marks")
plt.show()


plt.plot(df["Study_Hours"], df["Internal_Marks"], marker='o')
plt.title("Study Hours vs Internal Marks")
plt.xlabel("Study Hours")
plt.ylabel("Internal Marks")
plt.show()


print(df[["Attendance","Study_Hours","Internal_Marks"]].corr())
```

```python
import pandas as pd
import matplotlib.pyplot as plt


df = pd.read_excel("healthcare_dataset.xlsx")


print(df.head())
print(df.info())
print(df.describe())
print(df.isnull().sum())


plt.scatter(df["Sugar_Level"], df["BMI"])
plt.title("Sugar Level vs BMI")
plt.xlabel("Sugar Level")
plt.ylabel("BMI")
plt.show()


df["Disease_Status"].value_counts().plot(kind="bar")
plt.title("Disease Status")
plt.xlabel("Disease Status")
plt.ylabel("Number of Patients")
plt.show()


print(df[["Age", "Sugar_Level", "BP", "BMI"]].corr())
```

```python
import pandas as pd
import matplotlib.pyplot as plt


df = pd.read_excel("retail_dataset.xlsx")


print(df.head())
print(df.info())
print(df.describe())
print(df.isnull().sum())


plt.scatter(df["Price"], df["Revenue"])
plt.title("Price vs Revenue")
plt.xlabel("Price")
plt.ylabel("Revenue")
plt.show()


df.groupby("Category")["Revenue"].sum().plot(kind="bar")
plt.title("Revenue by Category")
plt.xlabel("Category")
plt.ylabel("Total Revenue")
plt.show()


print(df[["Price", "Quantity_Sold", "Discount", "Revenue"]].corr())
```

```python
import pandas as pd
import matplotlib.pyplot as plt


df = pd.read_excel("banking_dataset.xlsx")


print(df.head())
print(df.info())
print(df.describe())
print(df.isnull().sum())


plt.scatter(df["Income"], df["Credit_Score"])
plt.title("Income vs Credit Score")
plt.xlabel("Income")
plt.ylabel("Credit Score")
plt.show()


df["Loan_Status"].value_counts().plot(
    kind="pie",
    autopct="%1.1f%%",
    startangle=90
)
plt.title("Loan Approval Status")
plt.ylabel("")
plt.show()


print(df[["Income", "Credit_Score", "Loan_Amount"]].corr())
```

```python
import pandas as pd
import matplotlib.pyplot as plt


df = pd.read_excel("agriculture_dataset.xlsx")


print(df.head())
print(df.info())
print(df.describe())
print(df.isnull().sum())


plt.scatter(df["Rainfall_mm"], df["Crop_Yield_kg"])
plt.title("Rainfall vs Crop Yield")
plt.xlabel("Rainfall (mm)")
plt.ylabel("Crop Yield (kg)")
plt.show()


df["Soil_Type"].value_counts().plot(
    kind="pie",
    autopct="%1.1f%%",
    startangle=90
)
plt.title("Soil Type Distribution")
plt.ylabel("")
plt.show()


print(df[["Rainfall_mm", "Temperature", "Fertilizer_kg", "Crop_Yield_kg"]].corr())
```
