5. Smart Agriculture Data Monitoring

An agriculture research team collects temperature, humidity, soil moisture, and crop growth data from sensors. Explain how:

A. NumPy is used for numerical computation
B. Pandas is used for organizing sensor data
\
C. Scikit-Learn and visualization libraries support crop condition analysis

A. NumPy is used for numerical computation

```python
# Install library
!pip install openpyxl

# Import libraries
import pandas as pd
import numpy as np
from google.colab import files

# Upload Excel file
uploaded = files.upload()

# Read Excel file
df = pd.read_excel("agriculture_dataset.xlsx")

# Select only numeric columns
numeric_data = df.select_dtypes(include=np.number)

# NumPy Calculations
print("Average Values:")
print(np.mean(numeric_data, axis=0))

print("\nMaximum Values:")
print(np.max(numeric_data, axis=0))

print("\nMinimum Values:")
print(np.min(numeric_data, axis=0))

print("\nStandard Deviation:")
print(np.std(numeric_data, axis=0))
```

B. Pandas is used for organizing sensor data

```python
# Install library
!pip install openpyxl

import pandas as pd
from google.colab import files

# Upload Excel file
uploaded = files.upload()

# Read Excel file
df = pd.read_excel("agriculture_dataset.xlsx")

# Display dataset
print("First 5 Rows")
print(df.head())

print("\nDataset Information")
print(df.info())

print("\nMissing Values")
print(df.isnull().sum())

print("\nSummary Statistics")
print(df.describe())

# Display Temperature greater than 30 (optional)
if 'Temperature' in df.columns:
    print("\nTemperature > 30")
    print(df[df['Temperature'] > 30])
```

C. Scikit-Learn and visualization libraries support crop condition analysis


```python
import pandas as pd
import matplotlib.pyplot as plt

# Read Excel file
df = pd.read_excel("agriculture_dataset.xlsx")

# Count each soil type
soil_counts = df['Soil_Type'].value_counts()

# Pie Chart
plt.figure(figsize=(6,6))
plt.pie(soil_counts,
        labels=soil_counts.index,
        autopct='%1.1f%%',
        startangle=90)

plt.title("Distribution of Soil Types")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Read Excel file
df = pd.read_excel("agriculture_dataset.xlsx")

# Average Crop Yield for each Soil Type
avg_yield = df.groupby('Soil_Type')['Crop_Yield_kg'].mean()

# Bar Chart
plt.figure(figsize=(7,5))
avg_yield.plot(kind='bar')

plt.title("Average Crop Yield by Soil Type")
plt.xlabel("Soil Type")
plt.ylabel("Average Crop Yield (kg)")

plt.show()
```
